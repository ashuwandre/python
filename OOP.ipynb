```python
# Simple class

class Candidate:

    def __init__(self, name, skill, result):
        self.name = name
        self.skill = skill
        self.result = result

    def show(self):
        print(f"Name   : {self.name}")
        print(f"Skill  : {self.skill}")
        print(f"Result : {self.result}")


ashu = Candidate("Ashwini", "LLM", "Selected")

ashu.show()
```

```python
class Calculator:

    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divide(self, a, b):
        if b == 0:
            return "Cannot divide by zero"
        return a / b


calc = Calculator()

print("Addition:", calc.add(10, 5))
print("Subtraction:", calc.subtract(10, 5))
print("Multiplication:", calc.multiply(10, 5))
print("Division:", calc.divide(10, 5))
```

```python
#instance method  # classs methos # satatic methood

class Employee:

    company = "Amdocs"

    def __init__(self, name):
        self.name = name

    # Instance Method
    def employee_info(self):
        print(self.name, self.company)

    # Class Method
    @classmethod
    def change_company(cls, company):
        cls.company = company

    # Static Method
    @staticmethod
    def is_adult(age):
        return age >= 18


e1 = Employee("Ashwini")
e2 = Employee("Sachin")

e1.employee_info()
e2.employee_info()

Employee.change_company("Google")

e1.employee_info()
e2.employee_info()

print(Employee.is_adult(25))
```

```python
# inheritance 

class Animal:
    def sound(self):
        print('sound\n')
class Dog(Animal):
    def bark(self):
        print(f' dog ')
d = Dog()
d.sound()
d.bark()

```

```python
#polymorphism

class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


class Cat(Animal):
    def sound(self):
        print("Cat meows")


Dog().sound()
Cat().sound()
```

```python
#method overridding
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


class Cat(Animal):
    def sound(self):
        print("Cat meows")


Dog().sound()
Cat().sound()
```

```python
# method overloading

class Calculator:

    def add(self, a, b, c=0):
        return a + b + c


cal = Calculator()

print(cal.add(10, 20))
print(cal.add(10, 20, 30))
```

```python
# abstrctiosn

from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Bark")


dog = Dog()
dog.sound()
```

```python
# emcapsulation

class Employee:

    def __init__(self):
        self.name = "Ashwini"        # Public
        self._department = "AI"      # Protected
        self.__salary = 200000       # Private


emp = Employee()

print(emp.name)          # Public
print(emp._department)   # Protected (works, but not recommended)
print(emp.__salary)    # Error
```

```python
# decorator

def my_interview(func):
    def technical(name):
        
        print('passed')
        func(name)
        print('happy')
    return technical
@my_interview
def result(name):
    print(f'{name} selected')
result('ashwini')

```
