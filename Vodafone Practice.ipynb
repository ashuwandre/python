```python
# sorting
list1 = [33,5,6,7,2,1,3,4,6,7,99]

n = len(list1)

for i in range(n):
    for j in range(n - 1 - i):
        if list1[j] > list1[j + 1]:
            temp = list1[j]
            list1[j] = list1[j + 1]
            list1[j + 1] = temp

print(list1)
```

```python
# Reverse a Number
number = 123
reverse = 0
while number > 0:
    last_digit = number % 10
    reverse = reverse * 10 + last_digit
    number = number // 10
    
print(reverse)
```

```python
# Palindrome Number
number = 121
reverse = 0
temp = number

while number > 0:
    last_digit = number % 10
    reverse = reverse * 10 + last_digit
    number = number // 10

if reverse == temp:
    print("Yes")
else:
    print("no")
```

```python
# Armstrong Number

number = 125
temp = number
cube = 0

while number > 0:
    last_digit = number % 10
    cube = cube + last_digit **3
    number = number // 10
    
if temp == cube:
    print("yes")
else:
    print("No")
    
```
